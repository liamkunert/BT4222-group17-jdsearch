## 2. Locate and verify the data files

```python
from pathlib import Path

try:
    from google.colab import drive

    IN_COLAB = True
except ImportError:
    IN_COLAB = False

if IN_COLAB:
    drive.mount("/content/drive")
    DATA_DIR = Path("/content/drive/MyDrive/BT4222 Group Project/data")
    print("Running in Google Colab.")
else:
    local_candidates = [Path("../data"), Path("data")]
    DATA_DIR = next(
        (path.resolve() for path in local_candidates if path.is_dir()),
        Path("../data").resolve(),
    )
    print("Running locally.")

print(f"Data folder: {DATA_DIR}")
USER_BEHAVIOR_PATH = DATA_DIR / "user_behavior_data.txt"
PRODUCT_META_PATH = DATA_DIR / "product_meta_data.txt"

data_files = [USER_BEHAVIOR_PATH, PRODUCT_META_PATH]
missing_files = [path.name for path in data_files if not path.is_file()]

if missing_files:
    raise FileNotFoundError(
        f"Missing files: {missing_files}. "
        f"Place them in {DATA_DIR} before running this notebook."
    )

for path in data_files:
    size_gb = path.stat().st_size / (1024**3)
    print(f"Found {path.name} ({size_gb:.2f} GB)")
```
